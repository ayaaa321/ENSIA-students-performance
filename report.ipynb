## 🔬 Methodology Deep Dive

### Data Sources & Quality

| Dataset | Students | Description |
|---------|----------|-------------|

| `2y_final_clean_encoded.csv` | 101 | Year 2 cohort - balanced sample size |

### Why Our Preprocessing Approach Excels

#### 1. **Clean & Encoded Data Strategy**
We chose to work with pre-cleaned, encoded datasets rather than raw data for several compelling reasons:

- ✅ **Consistency**: Standardized encoding ensures categorical variables are handled uniformly
- ✅ **Reproducibility**: Clean data pipelines guarantee consistent results across runs
- ✅ **Efficiency**: Pre-processed data allows focus on modeling rather than data wrangling
- ✅ **Quality Assurance**: Missing values and outliers already addressed systematically

#### 2. **Target Variable Definition**
```
Pass: Yearly average ≥ 10/20
Fail: Yearly average < 10/20
```

**Why this threshold works**:
- Standard passing grade
- Creates a clear, binary classification problem
- Reflects real-world institutional requirements

## 🛠️ Alternative Approaches Considered & Why We Chose Our Path

### Approach Comparison

| Alternative Approach | Drawbacks | Our Chosen Approach | Advantages |
|---------------------|-----------|--------------------|-----------|
| Regression (predict exact grade) | Harder to interpret, unnecessary precision | **Binary Classification** | Clear pass/fail decision, actionable |
| Single algorithm | May miss optimal solution | **Multiple Algorithms** | Best model selection per dataset |
| Cross-year prediction | Data leakage risk, different contexts | **Year-specific Models** | Respects temporal boundaries, relevant features |
| Deep Learning | Overkill for tabular data, black box | **Traditional ML** | Interpretable, efficient, appropriate scale |

### Why Year-Specific Modeling is Superior

Rather than building one model to predict all years, we developed **separate models for each academic year**. This approach offers:

1. **Contextual Relevance**: Year 1 challenges differ from Year 3 challenges
2. **Feature Appropriateness**: Each year has unique course loads and assessment types
3. **No Data Leakage**: Future information never influences past predictions
4. **Targeted Insights**: Year-specific success factors can be identified

## 🤖 Classification Algorithms: Strategic Selection

### Algorithm Portfolio

#### 1. **Logistic Regression** (Baseline)
```
Why included:
• Establishes performance baseline
• Highly interpretable coefficients
• Works well with linearly separable data
• Fast training and prediction
```

#### 2. **Decision Tree** (Interpretability Champion)
```
Why included:
• Visual representation of decision logic
• Easy to explain to non-technical stakeholders
• Reveals feature importance naturally
• No feature scaling required
```

#### 3. **Random Forest** (Robust Performer)
```
Why included:
• Reduces overfitting through ensemble averaging
• Handles non-linear relationships
• Provides reliable feature importance
• Robust to outliers and noise
```

#### 4. **Gradient Boosting** (Performance Maximizer)
```
Why included:
• Often achieves best predictive accuracy
• Sequential error correction
• Handles complex feature interactions
• State-of-the-art for tabular data
```

### Why This Combination Works

This algorithm selection provides:
- **Coverage**: From simple linear to complex non-linear patterns
- **Comparison**: Fair evaluation across algorithm families
- **Flexibility**: Different models may excel for different years
- **Balance**: Trade-off between interpretability and performance

## 📊 Evaluation Framework

### Why Multiple Metrics Matter

In educational prediction, we use a comprehensive evaluation strategy:

| Metric | What It Tells Us | Educational Importance |
|--------|-----------------|----------------------|
| **Accuracy** | Overall correctness | General model quality |
| **Precision** | When we predict fail, how often correct? | Avoiding false alarms |
| **Recall** | What % of actual failures did we catch? | **Critical**: Missing at-risk students is costly |
| **F1-Score** | Balance of precision and recall | Overall classification quality |

### Cross-Validation: Ensuring Robustness

We employ **k-fold cross-validation** to:
- Maximize use of limited data (especially for Year 3's 12 students)
- Provide confidence intervals on performance
- Detect overfitting before deployment
