# Exploratory Data Analysis: Univariate Analysis

The objective of this notebook is to explore and understand the behavior of individual variables in the student performance dataset. 
Univariate analysis is conducted to examine the distribution, central tendency, variability, and potential outliers of each feature independently, without considering interactions between variables.

The analysis is performed on cleaned and imputed datasets produced by the preprocessing pipeline to ensure that observed patterns reflect meaningful student behavior rather than data quality artifacts.

The datasets used in this analysis are:
- `ensia_cleaned_1Y_grouped_imputed.csv`
- `ensia_cleaned_2Y_grouped_imputed.csv`
- `ensia_cleaned_3Y_grouped_imputed.csv`


```python
# Importing requireed libraries
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
import seaborn as sns
```

```python
#reading data from csv files 
data_1y = pd.read_csv('../data/outputs/grouped_imputed/ensia_cleaned_1Y_grouped_imputed.csv')
data_2y = pd.read_csv('../data/outputs/grouped_imputed/ensia_cleaned_2Y_grouped_imputed.csv')
data_3y = pd.read_csv('../data/outputs/grouped_imputed/ensia_cleaned_3Y_grouped_imputed.csv')
print("Data Loaded Successfully")
print("Data 1Y shape:", data_1y.shape)
data_1y.head()
print("Data 2Y shape:", data_2y.shape)
data_2y.head()
print("Data 3Y shape:", data_3y.shape)
data_3y.head()
```

# 1) Numerical Data

### a. First-Year Academic Performance (1Y) 

```python
numeric_cols = ['1Y_math_avg',
                '1Y_programming_avg', 
                '1Y_other_avg', 
                '1Y_all_modules_avg', 
                '1y average s1', 
                '1y average s2', 
                '1y average'
                ]

for col in numeric_cols: 
    if col in data_1y.columns:
        plt.figure(figsize=(12, 4))
        plt.subplot(1, 2, 1)
        sns.histplot(data_1y[col], kde=True, bins=30, color='skyblue')
        plt.title(f'Histogram of {col} - 1Y Data')
        plt.subplot(1, 2, 2)
        sns.boxplot(x=data_1y[col], color='pink')
        plt.title(f'Boxplot of {col} - 1Y Data')
        plt.show()
```

### b. Second-Year Academic Performance (2Y) 

```python
numeric_cols_2y = [
    '2Y_math_avg',
    '2Y_programming_avg',
    '2Y_other_avg',
    '2Y_all_modules_avg',
    '2y average s1',
    '2y average s2',
    '2y average',
]

for col in numeric_cols_2y: 
    if col in data_2y.columns:
        plt.figure(figsize=(12, 4))
        plt.subplot(1, 2, 1)
        sns.histplot(data_2y[col], kde=True, bins=30, color='skyblue')
        plt.title(f'Histogram of {col} - 2Y Data')
        plt.subplot(1, 2, 2)
        sns.boxplot(x=data_2y[col], color='pink')
        plt.title(f'Boxplot of {col} - 2Y Data')
        plt.show()
       
```

### c. Third-Year Academic Performance (3Y) 

```python
numeric_cols_3y = [
    '3Y_math_avg', 
    '3Y_programming_avg', 
    '3Y_other_avg', 
    '3Y_all_modules_avg',
    '3y average s1', 
    '3y average s2', 
    '3y average']

for col in numeric_cols_3y: 
    if col in data_3y.columns:
        plt.figure(figsize=(12, 4))
        plt.subplot(1, 2, 1)
        sns.histplot(data_3y[col], kde=True, bins=30, color='skyblue')
        plt.title(f'Histogram of {col} - 3Y Data')
        plt.subplot(1, 2, 2)
        sns.boxplot(x=data_3y[col], color='pink')
        plt.title(f'Boxplot of {col} - 3Y Data')
        plt.show()
```

### d. Student Motivation and Perception 

```python
motivation_cols = [
    'your level of motivation for joining ensia',
    'how much do you like the field of artificial intelligence',
    'how well do you handle stress and pressure'
    ]

all_motivation_data = pd.concat([data_1y[motivation_cols], data_2y[motivation_cols], data_3y[motivation_cols]], ignore_index=True)

for col in motivation_cols:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    
    sns.histplot(all_motivation_data[col], kde=True, bins=30, color='skyblue', ax=axes[0])
    axes[0].set_title(f'Histogram of {col} - All Years')
    
    sns.boxplot(x=all_motivation_data[col], color='pink', ax=axes[1])
    axes[1].set_title(f'Boxplot of {col} - All Years')

plt.show()
```

### c. Study Preferences 

```python
sns.set_style('whitegrid')

# Prepare the data
study_place = pd.concat([data_1y['where do you prefer to study'],
                         data_2y['where do you prefer to study'],
                         data_3y['where do you prefer to study']], ignore_index=True)
study_place_counts = study_place.value_counts()

study_style = pd.concat([data_1y['you prefer to study'],
                         data_2y['you prefer to study'],
                         data_3y['you prefer to study']], ignore_index=True)
study_style_counts = study_style.value_counts()

planner = pd.concat([data_1y['do you use a planer or a schedule in your revisions'],
                     data_2y['do you use a planer or a schedule in your revisions'],
                     data_3y['do you use a planer or a schedule in your revisions']], ignore_index=True)
planner_counts = planner.value_counts()

# Create subplots
fig, axes = plt.subplots(1, 3, figsize=(20,5))

# Plot 1: Preferred Study Place
sns.barplot(x=study_place_counts.index, y=study_place_counts.values,color='pink', ax=axes[0]    , edgecolor='black')
axes[0].set_title('Preferred Study Place')
axes[0].set_ylabel('Number of Students')
axes[0].set_xlabel('')
axes[0].tick_params(axis='x', rotation=45)

# Plot 2: Study Style
sns.barplot(x=study_style_counts.index, y=study_style_counts.values, color='skyblue', ax=axes[1], edgecolor='black')
axes[1].set_title('Study Style')
axes[1].set_ylabel('')
axes[1].set_xlabel('')
axes[1].tick_params(axis='x', rotation=30)

# Plot 3: Planner/Schedule Usage
sns.barplot(x=planner_counts.index, y=planner_counts.values, color='lightgreen', ax=axes[2], edgecolor='black')
axes[2].set_title('Planner/Schedule Usage')
axes[2].set_ylabel('')
axes[2].set_xlabel('')
axes[2].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

### d. Learning Methods

```python
import re 

learning_methods = pd.concat([
    data_1y['what learning methods do help you understand better'],
    data_2y['what learning methods do help you understand better'],
    data_3y['what learning methods do help you understand better']
], ignore_index=True)


# Split by both semicolons and commas (the data contains both delimiters)
all_methods = []
for response in learning_methods.dropna():
    # Split by both ; and ,
    methods = re.split('[;,]', response)
    # Strip whitespace and add to list
    all_methods.extend([method.strip() for method in methods if method.strip()])

# Count how many times each individual method appears
methods_counts = pd.Series(all_methods).value_counts()

print(f"Total unique methods found: {len(methods_counts)}")
print(f"\nTop 3 methods:")
print(methods_counts.head(3))
print("\n" + "="*50 + "\n")

# Create visualization
fig, ax = plt.subplots(figsize=(12, 7))
methods_counts.plot(kind='barh', ax=ax, color='lightblue', edgecolor='black')

# Styling
ax.set_xlabel('Number of Students', fontsize=12, fontweight='bold')
ax.set_ylabel('Learning Methods', fontsize=12, fontweight='bold')
ax.set_title('Individual Learning Methods - Frequency Count', 
             fontsize=14, fontweight='bold', pad=20)

# Add value labels on bars
for i, v in enumerate(methods_counts.values):
    ax.text(v + 0.5, i, str(v), va='center', fontweight='bold', fontsize=9)

# Add grid for readability
ax.grid(axis='x', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()

# Summary statistics
total_students = len(learning_methods.dropna())
print(f"Total students: {total_students}")
```

### g. Average Impact of Study Obstacle 

```python
import pandas as pd
import matplotlib.pyplot as plt

# Rank → score mapping
rank_to_score = {
    1: 8, 2: 7, 3: 6, 4: 5,
    5: 4, 6: 3, 7: 2, 8: 1
}

# Combine all years
obstacles_all = pd.concat(
    [data_1y[obstacle_cols], data_2y[obstacle_cols], data_3y[obstacle_cols]],
    ignore_index=True
)

impact_scores = {}

# Loop through ranking columns
for rank, col in enumerate(obstacle_cols, start=1):
    score = rank_to_score[rank]
    for obstacle in obstacles_all[col].dropna():
        impact_scores[obstacle] = impact_scores.get(obstacle, 0) + score

# Convert to DataFrame
impact_df = pd.DataFrame.from_dict(
    impact_scores, orient='index', columns=['Total Impact Score']
)

# Normalize by number of students
impact_df['Average Impact Score'] = (
    impact_df['Total Impact Score'] / len(obstacles_all)
)

# Sort
impact_df = impact_df.sort_values('Average Impact Score', ascending=True)

# Plot
plt.figure(figsize=(12,7))
plt.barh(
    impact_df.index,
    impact_df['Average Impact Score'],
    color='steelblue',
    edgecolor='black'
)

plt.xlabel('Average Impact Score')
plt.ylabel('Obstacle')
plt.title('Average Impact of Study Obstacles (Rank-Based)')
plt.grid(axis='x', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```
