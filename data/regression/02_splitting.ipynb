# Splitting (train/val/test)
This notebook runs `scripts/split_datasets.py` to produce `data/processed/splits/ensia_<version>_{train,val,test}.csv` and displays sizes.

```python
# Run the splitting script
import subprocess, os
py = os.path.join('.venv', 'Scripts', 'python.exe')
subprocess.run([py, 'scripts/split_datasets.py'], check=True)
```

```python
# Show counts for splits
import pandas as pd
from pathlib import Path
splits = Path('data/processed/splits')
for v in ['1y','2y','3y']:
    for t in ['train','val','test']:
        p = splits / f'ensia_{v}_{t}.csv'
        if p.exists():
            df = pd.read_csv(p)
            print(v.upper(), t, len(df))
```
